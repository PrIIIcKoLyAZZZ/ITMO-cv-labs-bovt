```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, random_split
from torchvision import datasets, transforms
import matplotlib.pyplot as plt
import numpy as np
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(device)

torch.manual_seed(42)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(42)

transform = transforms.ToTensor()

full_train_dataset = datasets.MNIST(
    root="./data",
    train=True,
    download=True,
    transform=transform
)

test_dataset = datasets.MNIST(
    root="./data",
    train=False,
    download=True,
    transform=transform
)

train_size = 50000
val_size = 10000
train_dataset, val_dataset = random_split(full_train_dataset, [train_size, val_size])

train_loader = DataLoader(train_dataset, batch_size=128, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=128, shuffle=False)
test_loader = DataLoader(test_dataset, batch_size=128, shuffle=False)
```

```python
class ReLUReg(nn.Module):
    def __init__(self, input_dim=28*28, hidden_dim=256, num_classes=10, dropout=0.3):
        super().__init__()
        self.net = nn.Sequential(
            nn.Flatten(),
            nn.Linear(input_dim, hidden_dim),
            nn.ReLU(),
            nn.Dropout(dropout),

            nn.Linear(hidden_dim, 128),
            nn.ReLU(),
            nn.Dropout(dropout),

            nn.Linear(128, num_classes)
        )

    def forward(self, x):
        return self.net(x)
```

```python
class ReLUNoReg(nn.Module):
    def __init__(self, input_dim=28*28, hidden_dim=256, num_classes=10):
        super().__init__()
        self.net = nn.Sequential(
            nn.Flatten(),
            nn.Linear(input_dim, hidden_dim),
            nn.ReLU(),

            nn.Linear(hidden_dim, 128),
            nn.ReLU(),

            nn.Linear(128, num_classes)
        )

    def forward(self, x):
        return self.net(x)
```

```python
class SigmoidReg(nn.Module):
    def __init__(self, input_dim=28*28, hidden_dim=256, num_classes=10, dropout=0.3):
        super().__init__()
        self.net = nn.Sequential(
            nn.Flatten(),
            nn.Linear(input_dim, hidden_dim),
            nn.Sigmoid(),
            nn.Dropout(dropout),

            nn.Linear(hidden_dim, 128),
            nn.Sigmoid(),
            nn.Dropout(dropout),

            nn.Linear(128, num_classes)
        )

    def forward(self, x):
        return self.net(x)
```

```python
class CNN(nn.Module):
    def __init__(self, num_classes=10):
        super().__init__()
        self.conv1 = nn.Conv2d(1, 16, kernel_size=3, padding=1)
        self.relu1 = nn.ReLU()
        self.pool1 = nn.MaxPool2d(2)

        self.conv2 = nn.Conv2d(16, 32, kernel_size=3, padding=1)
        self.relu2 = nn.ReLU()
        self.pool2 = nn.MaxPool2d(2)

        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(32 * 7 * 7, 128),
            nn.ReLU(),
            nn.Linear(128, num_classes)
        )

    def forward(self, x):
        x = self.pool1(self.relu1(self.conv1(x)))
        x = self.pool2(self.relu2(self.conv2(x)))
        x = self.classifier(x)
        return x
```

```python
def train_one_epoch(model, loader, optimizer, loss_fn, device):
    model.train()
    total_loss = 0.0
    correct = 0
    total = 0

    for X, y in loader:
        X, y = X.to(device), y.to(device)

        optimizer.zero_grad()
        logits = model(X)
        loss = loss_fn(logits, y)
        loss.backward()
        optimizer.step()

        total_loss += loss.item() * X.size(0)
        preds = torch.argmax(logits, dim=1)
        correct += (preds == y).sum().item()
        total += y.size(0)

    return total_loss / total, correct / total


@torch.no_grad()
def evaluate(model, loader, loss_fn, device):
    model.eval()
    total_loss = 0.0
    correct = 0
    total = 0

    for X, y in loader:
        X, y = X.to(device), y.to(device)
        logits = model(X)
        loss = loss_fn(logits, y)

        total_loss += loss.item() * X.size(0)
        preds = torch.argmax(logits, dim=1)
        correct += (preds == y).sum().item()
        total += y.size(0)

    return total_loss / total, correct / total


def fit(model, train_loader, val_loader, epochs=10, lr=1e-3, weight_decay=0.0):
    model = model.to(device)
    loss_fn = nn.CrossEntropyLoss()
    optimizer = optim.Adam(model.parameters(), lr=lr, weight_decay=weight_decay)

    history = {
        "train_loss": [],
        "train_acc": [],
        "val_loss": [],
        "val_acc": []
    }

    for epoch in range(epochs):
        train_loss, train_acc = train_one_epoch(model, train_loader, optimizer, loss_fn, device)
        val_loss, val_acc = evaluate(model, val_loader, loss_fn, device)

        history["train_loss"].append(train_loss)
        history["train_acc"].append(train_acc)
        history["val_loss"].append(val_loss)
        history["val_acc"].append(val_acc)

        print(
            f"Epoch {epoch+1}/{epochs} | "
            f"train_loss={train_loss:.4f} | train_acc={train_acc:.4f} | "
            f"val_loss={val_loss:.4f} | val_acc={val_acc:.4f}"
        )

    return model, history
```

```python
mlp_relu_reg = ReLUReg()
mlp_relu_reg, hist_relu_reg = fit(
    mlp_relu_reg,
    train_loader,
    val_loader,
    epochs=10,
    lr=1e-3,
    weight_decay=1e-4
)
```

```python
mlp_relu_noreg = ReLUNoReg()
mlp_relu_noreg, hist_relu_noreg = fit(
    mlp_relu_noreg,
    train_loader,
    val_loader,
    epochs=10,
    lr=1e-3,
    weight_decay=0.0
)
```

```python
mlp_sigmoid = SigmoidReg()
mlp_sigmoid, hist_sigmoid = fit(
    mlp_sigmoid,
    train_loader,
    val_loader,
    epochs=10,
    lr=1e-3,
    weight_decay=1e-4
)
```

```python
cnn = CNN(num_classes=10)
cnn, hist_cnn = fit(
    cnn,
    train_loader,
    val_loader,
    epochs=10,
    lr=1e-3
)
```

```python
def plot_history(history, title="Model"):
    epochs = range(1, len(history["train_loss"]) + 1)

    plt.figure(figsize=(12, 4))

    plt.subplot(1, 2, 1)
    plt.plot(epochs, history["train_loss"], label="train_loss")
    plt.plot(epochs, history["val_loss"], label="val_loss")
    plt.xlabel("Epoch")
    plt.ylabel("Loss")
    plt.title(f"{title}: Loss")
    plt.legend()

    plt.subplot(1, 2, 2)
    plt.plot(epochs, history["train_acc"], label="train_acc")
    plt.plot(epochs, history["val_acc"], label="val_acc")
    plt.xlabel("Epoch")
    plt.ylabel("Accuracy")
    plt.title(f"{title}: Accuracy")
    plt.legend()

    plt.tight_layout()
    plt.show()
```

```python
def compare_histories(histories, labels, metric_key, title, ylabel):
    plt.figure(figsize=(8, 5))
    for hist, label in zip(histories, labels):
        plt.plot(hist[metric_key], label=label)
    plt.xlabel("Epoch")
    plt.ylabel(ylabel)
    plt.title(title)
    plt.legend()
    plt.show()
```

```python
compare_histories(
    [hist_relu_reg, hist_relu_noreg, hist_sigmoid],
    ["ReLU + Reg", "ReLU no Reg", "Sigmoid + Reg"],
    "val_loss",
    "Validation Loss Comparison",
    "Loss"
)

compare_histories(
    [hist_relu_reg, hist_relu_noreg, hist_sigmoid],
    ["ReLU + Reg", "ReLU no Reg", "Sigmoid + Reg"],
    "val_acc",
    "Validation Accuracy Comparison",
    "Accuracy"
)
```

```python
loss_fn = nn.CrossEntropyLoss()

test_loss_reg, test_acc_reg = evaluate(mlp_relu_reg, test_loader, loss_fn, device)
test_loss_noreg, test_acc_noreg = evaluate(mlp_relu_noreg, test_loader, loss_fn, device)
test_loss_sigmoid, test_acc_sigmoid = evaluate(mlp_sigmoid, test_loader, loss_fn, device)

print("MLP + ReLU + Reg     :", test_loss_reg, test_acc_reg)
print("MLP + ReLU no Reg    :", test_loss_noreg, test_acc_noreg)
print("MLP + Sigmoid + Reg  :", test_loss_sigmoid, test_acc_sigmoid)
```

```python
@torch.no_grad()
def visualize_activations(model, dataset, index=0, device="cpu"):
    model.eval()

    image, label = dataset[index]
    x = image.unsqueeze(0).to(device)

    # Активации первых слоёв
    act1 = model.relu1(model.conv1(x))
    act1_pool = model.pool1(act1)

    act2 = model.relu2(model.conv2(act1_pool))

    image_np = image.squeeze().cpu().numpy()
    act1_np = act1.squeeze(0).cpu().numpy()
    act2_np = act2.squeeze(0).cpu().numpy()

    # Исходное изображение
    plt.figure(figsize=(3, 3))
    plt.imshow(image_np, cmap="gray")
    plt.title(f"Original image, label={label}")
    plt.axis("off")
    plt.show()

    # Карты первого слоя
    n1 = min(16, act1_np.shape[0])
    plt.figure(figsize=(12, 8))
    for i in range(n1):
        plt.subplot(4, 4, i + 1)
        plt.imshow(act1_np[i], cmap="gray")
        plt.title(f"conv1 map {i}")
        plt.axis("off")
    plt.tight_layout()
    plt.show()

    # Карты второго слоя
    n2 = min(16, act2_np.shape[0])
    plt.figure(figsize=(12, 8))
    for i in range(n2):
        plt.subplot(4, 4, i + 1)
        plt.imshow(act2_np[i], cmap="gray")
        plt.title(f"conv2 map {i}")
        plt.axis("off")
    plt.tight_layout()
    plt.show()
```

```python
visualize_activations(cnn, test_dataset, index=0, device=device)
```
