## Сгенерируйте случайный массив в NumPy. Продемонстрируйте на нём различные операции, которые рассматривались на практике

## Возьмите любое изображение из интернета, в котором присутствует хотя бы один объект. Воспользуйтесь функциями из OpenCV для выделения объекта-ов

```python
image_path = "data/apple.jpg"
image = cv2.imread(image_path)
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
gray = cv2.cvtColor(image, cv2.COLOR_RGB2GRAY)
blur = cv2.GaussianBlur(gray, (7, 7), 0)
```

```python
plt.imshow(blur)
```

```python
_, binary_mask = cv2.threshold(blur, 10, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)
plt.imshow(binary_mask)
```

```python
contours, _ = cv2.findContours(binary_mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

cnt = max(contours, key=cv2.contourArea)
x, y, w, h = cv2.boundingRect(cnt)
image = image.copy()
out = cv2.rectangle(image, (x, y), (x+w, y+h), (0, 255, 0), 2)

plt.imshow(cv2.cvtColor(out, cv2.COLOR_BGR2RGB))
plt.title("out")
plt.axis("off")

plt.show()
```

## Воспроизведите функцию rectangle из OpenCV с нуля. Разрешено использовать только NumPy для разработки из сторонних библиотек (не обязательно, чтобы параметры были те же самые, как в OpenCV, например thickness. Главное, чтобы прямоугольник было видно на изображении). Продемонстрируйте работу вашей функции на тестовом изображении

```python
def rectangle(img, x1, y1, x2, y2):
    out = img.copy()
    H, W = out.shape[:2]

    x1, x2 = sorted((int(x1), int(x2)))
    y1, y2 = sorted((int(y1), int(y2)))
    x1 = max(0, min(W-1, x1)); x2 = max(0, min(W-1, x2))
    y1 = max(0, min(H-1, y1)); y2 = max(0, min(H-1, y2))

    if out.ndim == 2:
        c = 255
        out[y1, x1:x2+1] = c
        out[y2, x1:x2+1] = c
        out[y1:y2+1, x1] = c
        out[y1:y2+1, x2] = c
    else:
        c = np.array([255, 255, 255], dtype=out.dtype)
        out[y1, x1:x2+1, :] = c
        out[y2, x1:x2+1, :] = c
        out[y1:y2+1, x1, :] = c
        out[y1:y2+1, x2, :] = c

    return out
```

```python
H, W = 200, 300
img = np.zeros((H, W, 3), dtype=np.uint8)
img[30:170, 50:250] = np.zeros(3, dtype=np.uint8)
img[30:170, 50:250] = 255

out = rectangle(img, 40, 20, 260, 180)

plt.figure(figsize=(8, 3))
plt.subplot(1, 2, 1); plt.imshow(img); plt.title("original"); plt.axis("off")
plt.subplot(1, 2, 2); plt.imshow(out); plt.title("rect"); plt.axis("off")
plt.tight_layout()
plt.show()
```
